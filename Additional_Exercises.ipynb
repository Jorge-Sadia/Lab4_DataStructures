```python
# Appendix Exercise 1: Engineering Test Data Organizer
LAST_NAME = "Sadia"
SEED_NUM = 0
FAVORITE_ARTIST = "Taylor Swift"

sources = [f"{LAST_NAME}_T1", f"{LAST_NAME}_T2", f"{LAST_NAME}_T3", f"{LAST_NAME}_T4", f"{LAST_NAME}_T5"]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
val_modifier = sum(ord(char) for char in FAVORITE_ARTIST) % 50

generated_test_data = []
for idx, src in enumerate(sources):
    cat = categories[idx]
    val = round(float(SEED_NUM + val_modifier + (idx * 5.5)), 2)
    generated_test_data.append({"Source": src, "Category": cat, "Value": val})

# Requirements: Include one repeated record and multiple measurements under two categories
generated_test_data.append({"Source": f"{LAST_NAME}_T1", "Category": "TEMPERATURE", "Value": round(float(SEED_NUM + val_modifier), 2)})
generated_test_data.append({"Source": f"{LAST_NAME}_T1", "Category": "VOLTAGE", "Value": round(float(SEED_NUM + val_modifier + 12.3), 2)})
generated_test_data.append({"Source": f"{LAST_NAME}_T2", "Category": "TEMPERATURE", "Value": round(float(SEED_NUM + val_modifier + 8.7), 2)})

# Identify distinct and repeated entries using Sets
seen_records = set()
repeated_records = []
distinct_records = []

for record in generated_test_data:
    record_tuple = (record["Source"], record["Category"], record["Value"])
    if record_tuple in seen_records:
        repeated_records.append(record)
    else:
        seen_records.add(record_tuple)
        distinct_records.append(record)

values = [r["Value"] for r in generated_test_data]

print("=== 1. GENERATED TEST DATA ===")
for r in generated_test_data: print(r)
print("\n=== 2. DISTINCT RECORDS ===")
for r in distinct_records: print(r)
print("\n=== 3. REPEATED RECORDS ===")
for r in repeated_records: print(r)
print(f"\n=== 4. NUMERICAL SUMMARIES ===\nMax: {max(values)} | Min: {min(values)} | Avg: {round(sum(values)/len(values), 2)}")
```

```python
# Appendix Exercise 2: Student Performance Data Analyzer
LAST_NAME = "Sadia"
SEED_NUM = 0

subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]
student_labels = [f"{LAST_NAME}_S1", f"{LAST_NAME}_S2", f"{LAST_NAME}_S3", f"{LAST_NAME}_S4", f"{LAST_NAME}_S5"]

raw_scores = [
    {"Student": f"{LAST_NAME}_S1", "Subject": "PROGRAMMING", "Score": 85},
    {"Student": f"{LAST_NAME}_S2", "Subject": "MATHEMATICS", "Score": 92},
    {"Student": f"{LAST_NAME}_S3", "Subject": "ELECTRONICS", "Score": 74},
    {"Student": f"{LAST_NAME}_S4", "Subject": "PHYSICS", "Score": 88},
    {"Student": f"{LAST_NAME}_S5", "Subject": "ENGLISH", "Score": 65},
    # Requirements: Include one repeated record and one incomplete record
    {"Student": f"{LAST_NAME}_S1", "Subject": "PROGRAMMING", "Score": 85},
    {"Student": f"{LAST_NAME}_S2", "Subject": "ELECTRONICS", "Score": None}
]

processed_records = []
repeated_records = []
incomplete_records = []
seen_pairs = set()

for r in raw_scores:
    score = r["Score"]
    pair = (r["Student"], r["Subject"])
    
    # Identify repeated or incomplete
    if pair in seen_pairs: repeated_records.append(r)
    else: seen_pairs.add(pair)
    if score is None: incomplete_records.append(r)
        
    # Categorize performance
    if score is None: cat = "INCOMPLETE"
    elif score >= 90: cat = "EXCELLENT"
    elif score >= 80: cat = "GOOD"
    elif score >= 70: cat = "PASSING"
    else: cat = "FAILING"
    
    rec_copy = r.copy()
    rec_copy["Category"] = cat
    processed_records.append(rec_copy)

print("=== 1. GENERATED PERFORMANCE RECORDS ===")
for r in processed_records: print(r)
print("\n=== 2. REPEATED RECORDS ===")
for r in repeated_records: print(r)
print("\n=== 3. INCOMPLETE RECORDS ===")
for r in incomplete_records: print(r)
```

```python
# Appendix Exercise 3: Transaction History Analyzer
LAST_NAME = "Sadia"
SEED_NUM = 0

customers = ["C1", "C2", "C3", "C4", "C5"]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 50, "PEN": 15, "USB": 350, "HEADPHONES": 500, "CALCULATOR": 250}

transactions = [
    {"Customer": f"{LAST_NAME}_C1", "Product": "NOTEBOOK", "Qty": 3},
    {"Customer": f"{LAST_NAME}_C2", "Product": "PEN", "Qty": 5},
    {"Customer": f"{LAST_NAME}_C3", "Product": "USB", "Qty": 1},
    {"Customer": f"{LAST_NAME}_C4", "Product": "HEADPHONES", "Qty": 2},
    # Requirement: Include at least one repeated customer-product purchase
    {"Customer": f"{LAST_NAME}_C1", "Product": "NOTEBOOK", "Qty": 3}
]

print("=== CONSOLIDATED TRANSACTION REPORT ===")
grand_total = 0
for tx in transactions:
    item_price = prices[tx["Product"]]
    total_cost = item_price * tx["Qty"]
    grand_total += total_cost
    print(f"Customer: {tx['Customer']} | Item: {tx['Product']} | Qty: {tx['Qty']} | Total Value: Php {total_cost}")

print(f"\nOVERALL TRANSACTION VALUE SUMMARY (GRAND TOTAL): Php {grand_total}")
```
